# Mathematical Optimisations

Reducing the number of calculations in a piece of code will typically reduce the runtime. Many codes spend a significant portion of their time performing maths and replacing long-winded calculations with shorter calculations which are algebraically identical can reduce the runtime significantly. For instance, consider the three examples below:

### Example 1

````python
%load_ext line_profiler

def function1(value):
  thousand_times=0

  for i in range(1000):
    thousand_times = thousand_times + value

  return thousand_times

%lprun -f function1 print(function1(10000))
````

We can replace the sum with a multiplication:

````python
%load_ext line_profiler

def function1(value):
  return 1000 * value

%lprun -f function1 print(function1(10000))
````

### Example 2

````python
%load_ext line_profiler

def function2(r, n):
  geometric_sum=0

  for i in range(0, n):
    geometric_sum = geometric_sum + r ** i

  return geometric_sum

%lprun -f function2 print(function2(0.99, 1000))
````

We can use the geometric sum equation:

$\sum\limits_{i=0}^{n-1}r^{i} = \frac{1-r^{n}}{1-r}$

to simplify the sum into a single expression:

````python
%load_ext line_profiler

def function2(r, n):
  return (1 - r ** n) / (1 - r)

%lprun -f function2 print(function2(0.99, 1000))
````

### Example 3

````python
import math

%load_ext line_profiler

def function3(n):
  trig_sum = 0

  for i in range(n):
    trig_sum = trig_sum + math.sqrt(1 - math.cos(i/n) ** 2)

  return trig_sum

%lprun -f function3 print(function3(1000))
````

We can use the trigonometric identity:

$\sin(x) = \sqrt{1-\cos^{2}(x)}$

This reduces the complexity of the expression evaluated in the loop:

````python
import math

%load_ext line_profiler

def function3(n):
  trig_sum = 0

  for i in range(n):
    trig_sum = trig_sum + math.sin(i / n)

  return trig_sum

%lprun -f function3 print(function3(1000))
````

## Cost of Mathematical Functions

A common way to optimise code is to find the least "expensive" (i.e. lowest running time) combination of mathematical operations which return the desired value. This may be through:

- Using standard mathematical definitions to simplify the algebraic expression being evaluated
- Rearranging the algebraic expression being solved for to reduce the number of more expensive operations being carried out

Mathematical operations will be carried out on your computer by passing relevant data to the [Arithmetic Logic Unit](https://en.wikipedia.org/wiki/Arithmetic_logic_unit) (ALU) on your CPU (or GPU). ALUs typically contain circuits specifically designed to carry out fundamental operations (such as addition). More complex functions such as `sin` or `log` are often implemented through highly optimised code written in C (such as [`libm`](https://docs.amd.com/r/2021.1-English/ug1137-zynq-ultrascale-mpsoc-swdev/The-C-Standard-Library-Mathematical-Functions-libm))which uses a combinations of these fundamental operations to perform more complex operations.

Every mathematical operator and function will be executed by using one or more operations on your computer's ALU. These operations will take time to execute on your computer and the number and complexity of operations required will affect the running time. Thus, some operations are more "expensive" to perform than others.

````python
import math

%load_ext line_profiler

n_repetitions = 100000

def mathematical_operators(a,b):
  for i in range(1, n_repetitions):
    a
  for i in range(1, n_repetitions):
    a == b
  for i in range(1, n_repetitions):
    a < b
  for i in range(1, n_repetitions):
    a + b
  for i in range(1, n_repetitions):
    a - b
  for i in range(1, n_repetitions):
    a * b
  for i in range(1, n_repetitions):
    a / b
  for i in range(1, n_repetitions):
    a ** b
  for i in range(1, n_repetitions):
    math.log(a)
  for i in range(1, n_repetitions):
    math.log(a, b)
  for i in range(1, n_repetitions):
    math.sqrt(a)
  for i in range(1, n_repetitions):
    math.exp(a)
  for i in range(1, n_repetitions):
    math.sin(a)
  for i in range(1, n_repetitions):
    math.acos(a)
  for i in range(1, n_repetitions):
    math.tan(a)
  for i in range(1, n_repetitions):
    math.atan(a)
  for i in range(1, n_repetitions):
    math.cosh(a)
  for i in range(1, n_repetitions):
    math.asinh(a)
  for i in range(1, n_repetitions):
    math.tanh(a)
  for i in range(1, n_repetitions):
    math.atanh(a)
  for i in range(1, n_repetitions):
    math.ceil(a)
  for i in range(1, n_repetitions):
    math.gamma(a)
  for i in range(1, n_repetitions):
    math.factorial(4)
  for i in range(1, n_repetitions):
    math.factorial(40)

%lprun -f mathematical_operators mathematical_operators(0.96, 4.12)
````

This means there is sometimes a gain to be made by reformulating your maths to use more fundamental, cheaper operations where possible. However, the difference is low enough that reducing the overall number of operations is often more important.

## More Specific vs Less Specific Functions

Note as well that when functions should have the same result, the more specialised function will normally have the shorter run-time. For example, ```math.log10(a)``` will be faster than ```math.log(a, 10)```. This is because these more specialised versions can employ shortcuts which their more generally applicable counterparts cannot:

````python
import math

%load_ext line_profiler

def log_function_comparisons():
  for i in range(1, 10000):
    math.log10(3.5)

  for i in range(1, 10000):
    math.log(3.5, 10)

%lprun -f log_function_comparisons log_function_comparisons()
````

# Use Built-In Functions

If there's a built-in version of a function available, or one from a standard libarary or package you can import, it is often significantly faster than anything you could write. This is partially because most functions in modules such as the `math` module actually execute compiled code written in the "C" language which is much faster than Python and partially because the authors of Python's intrinsic functions and modules are very good programmers. As a result, if there is an intrinsic version of a function, it is almost always preferable to use it rather than write your own.

````python
%load_ext line_profiler

# Custom implementation of a function which sums 
def custom_sum(lst):
    result = 0
    for item in lst:
        result = result + item

    return result

def sum_comparison(lst, n_repetitions):
    for i in range(n_repetitions):
        custom_result = custom_sum(lst)

    for i in range(n_repetitions):
        builtin_result = sum(lst)

# Create a list of 1000 elements
lst = [1, 2, 3, 4, 5] * 200
n_repetitions = 1000

print("Custom implementations:")
%lprun -f sum_comparison sum_comparison(lst, n_repetitions)
````
